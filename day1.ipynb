# Data Visualization

- Line Chart: `plt.plot()`
- Bar Chart: `plt.bar()`
- Scatter Plot: `plt.scatter()`
- Histogram: `plt.hist()`
- Box Plot: `sns.boxplot()`
- Pie Chart: `plt.pie()`

```python
import matplotlib.pyplot as plt
```

```python
months=["jan","feb","march","april","may"]
sales=[100,120,150,130,180]

plt.plot(months,sales,marker="*")

plt.title("monthly sales")
plt.xlabel("month")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

```python
subjects=["maths","python","sql","ai"]
marks=[75,85,70,90]
plt.bar(subjects,marks,color="orange",edgecolor="black")
plt.title("student marks")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
subjects=["maths","python","sql","ai"]
marks=[75,85,70,90]
plt.bar(subjects,marks,color="orange",edgecolor="skyblue",alpha=0.5,hatch="*")
plt.title("student marks")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
subjects=["maths","python","sql","ai"]
marks=[75,85,70,90]
plt.bar(subjects,marks,color="red",edgecolor="skyblue",alpha=0.5,hatch="/")
plt.title("student marks")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
x=["maths","python","sql","ai"]
y=[75,85,70,90]
plt.plot(x,y,linestyle="--")
plt.title("student marks")
plt.xlabel("subjects")
plt.ylabel("marks")
plt.grid(True)
plt.show()
```

```python
study_hours=[1,2,3,4,5,6]
marks=[45,50,60,65,75,85]
plt.scatter(study_hours,marks,s=10)
plt.title("Student Study hours vs Marks")
plt.xlabel("hours studied")
plt.ylabel("mark")

plt.show()

plt.scatter(study_hours,marks,color="green",edgecolor="yellow")
plt.title("Student Study hours vs Marks")
plt.xlabel("hours studied")
plt.ylabel("mark")

plt.show()
```

```python
marks=[45,50,55,60,62,65,68,70,72,75,78,80,82]

plt.hist(marks,bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("No of Students")

plt.show()
```

```python
subjects=["python","sql","ai","powerbi","tableau"]
hours=[10,6,8,5,6]
plt.pie(hours, labels=subjects,autopct="%1.1f%%")
plt.title("study time distribution")
```
